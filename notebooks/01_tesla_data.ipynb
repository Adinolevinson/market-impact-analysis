```python
import pandas as pd

spy = pd.read_csv("../data/raw/SPY_stock_2019_2026.csv")

spy["Date"] = pd.to_datetime(spy["Date"])
```

```python
window = get_event_window("TSLA_011", events, tesla)
```

```python
# Add SPY closing prices for the same trading dates
window = window.merge(
    spy[["Date", "Close"]],
    on="Date",
    how="left",
    suffixes=("", "_SPY")
)

# Rename the SPY closing-price column
window = window.rename(
    columns={"Close_SPY": "SPY_Close"}
)

# Get SPY's Day 0 closing price
spy_day_0_close = window.loc[
    window["day"] == 0,
    "SPY_Close"
].iloc[0]

# Calculate SPY return from the same Day 0 baseline
window["SPY_return_from_day_0"] = (
    (window["SPY_Close"] / spy_day_0_close) - 1
) * 100

# Calculate Tesla's return relative to the market
window["abnormal_return"] = (
    window["return_from_day_0"]
    - window["SPY_return_from_day_0"]
)

window
```
