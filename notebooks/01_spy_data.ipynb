```python
import yfinance as yf
import pandas as pd

# Download SPY historical data
spy = yf.download(
    "SPY",
    start="2019-01-01",
    end="2026-01-01",
    auto_adjust=True
)

# Remove the ticker level if yfinance returns MultiIndex columns
if isinstance(spy.columns, pd.MultiIndex):
    spy.columns = spy.columns.get_level_values(0)

# Move Date from the index into a normal column
spy = spy.reset_index()

# Make sure Date is stored as datetime
spy["Date"] = pd.to_datetime(spy["Date"])

# Keep the same columns as the Tesla dataset
spy = spy[
    ["Date", "Close", "High", "Low", "Open", "Volume"]
]

# Save it once in raw data
spy.to_csv(
    "../data/raw/SPY_stock_2019_2026.csv",
    index=False
)

spy.head()
```
